**Activity 1: Student Info Formatter (Version 2)**

```python
# Asks for student information
first_name = input("Enter first name: ").strip()
last_name = input("Enter last name: ").strip()
section = input("Enter section: ").strip()
favorite_subject = input("Enter favorite subject: ").strip()

# Format the information
full_name = f"{first_name}  {last_name}"
proper_name = full_name.title()
upper_name = full_name.upper()
upper_section = section.upper()
proper_subject = favorite_subject.title()
char_count = len(full_name)

# Display the formatted information
print("\nFormatted Output:")
print(f"Name: {proper_name}")
print(f"Name Code: {upper_name}")
print(f"Section: {upper_section}")
print(f"Favorite Subject: {proper_subject}")
print(f"Character Count: {char_count}")
```

Activity 2: Deadline Date Formatter (Version 2)

```python
from datetime import datetime

event = input("Enter event title: ")
year = int(input("Enter year: "))
month = int(input("Enter month: "))
day = int(input("Enter day: "))

date = datetime(year, month, day)

print("\nFormatted Output:")
print("Event:", event.upper())
print("Date:", date.strftime)
```
